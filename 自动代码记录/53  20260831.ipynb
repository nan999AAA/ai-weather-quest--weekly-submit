```bash
# 1 复制数据文件夹
#!/bin/bash

# ==========================================
# 复制并重命名次季节预测目录
#
# 功能：
# 1. 复制：
#    /Users/zhangnan/日常文件/ai/每周下载和预测/52 20260824/数据
#
# 2. 新目录命名为：
#    53 20260831
#
# 3. 所有【非 nc 数据文件】的：
#    文件名、目录名、文件内容：
#
#    52 20260824 -> 53 20260831
#    20260824   -> 20260831
#
# 4. download 文件夹只复制目录结构
#    不复制其中的数据文件
# ==========================================

set -e

# 原目录
SRC="/Users/zhangnan/日常文件/ai/每周下载和预测/52 20260824/数据"

# 目标目录
DST="/Users/zhangnan/日常文件/ai/每周下载和预测/53 20260831/数据"

# 创建目标目录
mkdir -p "$DST"

echo "===================================="
echo "开始复制目录..."
echo "===================================="

# ==========================================
# 复制文件
# download 目录不复制数据文件
# ==========================================
rsync -av \
    --exclude='download/*.nc' \
    --exclude='download/*.grib' \
    --exclude='download/*.zip' \
    --exclude='download/*.tar' \
    --exclude='download/*.gz' \
    --exclude='download/*.bz2' \
    --exclude='download/*.h5' \
    --exclude='download/*.hdf5' \
    --exclude='download/*.csv' \
    --exclude='download/*.npy' \
    --exclude='download/*.npz' \
    --exclude='data/*.nc' \
    --exclude='data/*.grib' \
    --exclude='data/*.zip' \
    --exclude='data/*.tar' \
    --exclude='data/*.gz' \
    --exclude='data/*.bz2' \
    --exclude='data/*.h5' \
    --exclude='data/*.hdf5' \
    --exclude='data/*.csv' \
    --exclude='data/*.npy' \
    --exclude='data/*.npz' \
    "$SRC/" "$DST/"

echo "===================================="
echo "开始重命名目录..."
echo "===================================="

# ==========================================
# 只重命名目录
# ==========================================
find "$DST" -depth -type d | while read -r dir; do

    new_dir=$(echo "$dir" | sed \
        -e 's/52 20260824/53 20260831/g' \
        -e 's/20260824/20260831/g')

    if [ "$dir" != "$new_dir" ]; then
        mv "$dir" "$new_dir"
    fi
done

echo "===================================="
echo "开始重命名文件（跳过 .nc）..."
echo "===================================="

# ==========================================
# 非 nc 文件重命名
# ==========================================
find "$DST" -type f ! -name "*.nc" | while read -r file; do

    new_file=$(echo "$file" | sed \
        -e 's/52 20260824/53 20260831/g' \
        -e 's/20260824/20260831/g')

    if [ "$file" != "$new_file" ]; then
        mv "$file" "$new_file"
    fi
done

echo "===================================="
echo "开始修改文件内容（跳过 .nc）..."
echo "===================================="

# ==========================================
# 修改文本内容
# 跳过 nc 文件
# ==========================================
find "$DST" -type f ! -name "*.nc" \( \
    -name "*.ipynb" -o \
    -name "*.py" -o \
    -name "*.sh" -o \
    -name "*.txt" -o \
    -name "*.md" -o \
    -name "*.json" -o \
    -name "*.yaml" -o \
    -name "*.yml" \
\) | while read -r file; do

    sed -i '' \
        -e 's/52 20260824/53 20260831/g' \
        -e 's/20260824/20260831/g' \
        -e 's/2026-08-24/2026-08-31/g' \
        "$file"

done

echo "===================================="
echo "完成！"
echo "目标目录："
echo "$DST"
echo "===================================="
```

```bash
# 2 复制AZN的脚本并重命名
#!/bin/bash

# =========================================================
# 复制次季节预测脚本（仅 .ipynb）
#
# 功能：
# 1. 从：
#    /Users/zhangnan/日常文件/ai/每周下载和预测/52 20260824
#
# 2. 复制以下目录中的 .ipynb 文件：
#    - 20260824-mslp
#    - 20260824-tas
#    - 20260824-tp
#
# 3. 不复制：
#    - .ipynb_checkpoints
#    - 非 .ipynb 文件
#
# 4. 仅保留【包含 ipynb 文件】的目录结构
#    空目录不会复制
#
# 5. 所有：
#    目录名
#    文件名
#    ipynb 文件内容
#
#    执行：
#
#    52 20260824 -> 53 20260831
#    20260824   -> 20260831
#    2026-08-24 -> 2026-08-31
#
# =========================================================

set -e

# =========================================================
# 原目录
# =========================================================
SRC_ROOT="/Users/zhangnan/日常文件/ai/每周下载和预测/52 20260824"

# =========================================================
# 目标目录
# =========================================================
DST_ROOT="/Users/zhangnan/日常文件/ai/每周下载和预测/53 20260831"

# =========================================================
# 创建目标目录
# =========================================================
mkdir -p "$DST_ROOT"

echo "===================================="
echo "开始复制 .ipynb 文件..."
echo "===================================="

# =========================================================
# 需要复制的目录
# =========================================================
DIRS=(
    "20260824-mslp"
    "20260824-tas"
    "20260824-tp"
)

# =========================================================
# 复制 ipynb 文件
# 不复制 .ipynb_checkpoints
# 仅复制含 ipynb 的目录结构
# =========================================================
for dir in "${DIRS[@]}"; do

    SRC_DIR="$SRC_ROOT/$dir"

    find "$SRC_DIR" \
        -type f \
        -name "*.ipynb" \
        ! -path "*/.ipynb_checkpoints/*" \
    | while read -r file; do

        # 相对路径
        rel_path="${file#$SRC_ROOT/}"

        # 目标文件路径
        dst_file="$DST_ROOT/$rel_path"

        # 创建目标目录
        mkdir -p "$(dirname "$dst_file")"

        # 复制文件
        cp "$file" "$dst_file"

    done

done

echo "===================================="
echo "开始重命名目录..."
echo "===================================="

# =========================================================
# 重命名目录
# =========================================================
find "$DST_ROOT" -depth -type d | while read -r dir; do

    new_dir=$(echo "$dir" | sed \
        -e 's/52 20260824/53 20260831/g' \
        -e 's/20260824/20260831/g')

    if [ "$dir" != "$new_dir" ]; then

        mkdir -p "$(dirname "$new_dir")"

        mv "$dir" "$new_dir"

    fi

done

echo "===================================="
echo "开始重命名文件..."
echo "===================================="

# =========================================================
# 重命名 ipynb 文件
# =========================================================
find "$DST_ROOT" -type f -name "*.ipynb" | while read -r file; do

    new_file=$(echo "$file" | sed \
        -e 's/52 20260824/53 20260831/g' \
        -e 's/20260824/20260831/g')

    if [ "$file" != "$new_file" ]; then

        mv "$file" "$new_file"

    fi

done

echo "===================================="
echo "开始修改 ipynb 文件内容..."
echo "===================================="

# =========================================================
# 修改 ipynb 文件内容
# =========================================================
find "$DST_ROOT" -type f -name "*.ipynb" | while read -r file; do

    sed -i '' \
        -e 's/52 20260824/53 20260831/g' \
        -e 's/20260824/20260831/g' \
        -e 's/2026-08-24/2026-08-31/g' \
        "$file"

done

echo "===================================="
echo "完成！"
echo "目标目录："
echo "$DST_ROOT"
echo "===================================="
```
